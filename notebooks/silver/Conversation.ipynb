```python
from pyspark.sql.functions import *

bronze_df = (
    spark.table("datastore.bronze.conversation")
    .select("participants", "conversationId")
    .filter(col("conversationId") == "40af493f-a66a-4723-8604-6b5cbeba7888")
)
                                                  
                                                 
```

```python
bronze_df = (
    bronze_df
    .withColumn("segment", explode(col("participants")))
    .select(
        "conversationId",
        col("segment.purpose").alias("purpose")
    )
    .dropDuplicates(["conversationId", "purpose"])
)
```
